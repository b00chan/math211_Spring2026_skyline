```python
def longest_streak(row):
    max_value = None
    max_count = 0
    
    current_value = None
    current_count = 0
    
    for value in row:
        #print(value)
        if value == current_value:
            #print("cv",current_value)
            current_count += 1
        else:
            current_value = value
            current_count = 1
        
        if current_count > max_count:
            max_count = current_count
            max_value = current_value
    
    return make_array(max_value, max_count)
```

```python
def detect_fraud(table_with_single_row):
    first_trial = table_with_single_row.select(np.arange(1,101)).row(0)
    second_trial = table_with_single_row.select(np.arange(101,201)).row(0)

    streak_length_first_trial = longest_streak(first_trial).item(1)
    streak_length_second_trial = longest_streak(second_trial).item(1)

    head_or_tail_first = longest_streak(first_trial).item(0)
    head_or_tail_second = longest_streak(second_trial).item(0)

    if streak_length_first_trial > streak_length_second_trial:
        return "Trial 1 produced the longest chain of",head_or_tail_first,"which appeared",streak_length_first_trial,"times!"
    elif streak_length_first_trial < streak_length_second_trial:
        return "Trial 1 produced the longest chain of",head_or_tail_second,"which appeared",streak_length_second_trial,"times!"
    else: 
        return "BRUH, the tests results are inconclusive"
    
```

```python
def determine_bet(table,name):
    single_row_table = table.where('Group',are.containing(name))
    return detect_fraud(single_row_table)
    
    
```
